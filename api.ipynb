```python
import requests
import pandas as pd
import json
import os
```

```python
# URL REST API
API_URL = "https://portaldata.kkp.go.id/api/portal/list_kawasan_bahari"

# Mengirim HTTP GET request
response = requests.get(API_URL, timeout=30, params={'limit': 200})

# Memastikan request berhasil
response.raise_for_status()

# Mengubah respons menjadi JSON
result = response.json()

# Mengambil informasi dataset
total_data = result["total_count"]
data = result["data"]

# Mengubah data menjadi DataFrame
df_luwu = pd.DataFrame(data)

# Menampilkan informasi dataset
print("Total Data:", total_data)
print("Jumlah Data Berhasil Diambil:", len(df_luwu))

display(df_luwu)
output_path = "data/raw/luwu_timur.csv"
print(output_path);

# Simpan data Raw
df_luwu.to_csv(output_path, index=False)
print("Dataset Raw berhasil disimpan:", output_path)
```

```python
# Kolom yang akan dihapus
kolom_hapus = [
    "created_on",
    "created_by",
    "updated_on",
    "updated_by"
]

# Menghapus kolom
df_cropped = df_luwu.drop(columns=kolom_hapus, errors="ignore")
df_aggregat = df_cropped.groupby(
    ['tahun', 'provinsi'],
    as_index=False
).agg({
    'kabupaten': 'first',
    'kecamatan': 'first',
    'desa': 'first',
    'kelompok_pengelola': 'first',
    'lat': 'first',
    'long': 'first'
})

# Menampilkan hasil preprocessing
display(df_cropped)
display(df_aggregat)

# Memeriksa shape terbaru
print("Shape setelah preprocessing:", df_cropped.shape)
```

```python
output_path1 = "data/process/list_kawasan_bahari_cropped.csv"
output_path2 = "data/process/list_kawasan_bahari_order_by.csv"
# print(output_path);

# Simpan data Process
df_cropped.to_csv(output_path1, index=False)
df_aggregat.to_csv(output_path2, index=False)
print("Dataset berhasil disimpan:", output_path1)
print("Dataset berhasil disimpan:", output_path2)
```
