```python
import requests
import pandas as pd
from bs4 import BeautifulSoup
from urllib.parse import urljoin
```

```python
# URL target
BASE_URL = "https://id.wikipedia.org"
TARGET_URL = "https://id.wikipedia.org/wiki/Pariwisata_di_Bali"

# Header browser
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 Chrome/120.0.0.0 Safari/537.36"
}

# Request halaman
response = requests.get(
    TARGET_URL,
    headers=headers,
    timeout=30
)

print("Status Code:", response.status_code)
response.raise_for_status()

html = response.text

# Parsing HTML
soup = BeautifulSoup(html, "lxml")
```

```python
# Mencari div secara spesifik
section = soup.find(
    "section",
    attrs={"data-mw-section-id": "8"}
)

if section:
    print("Section ditemukan!")
    print(section.prettify()[:3000])
else:
    print("Section dengan data-mw-section-id='8' tidak ditemukan!")
```

```python
data = []

if section:
    links = section.find_all(
        "a",
        attrs={"rel": "mw:WikiLink"}
    )

    for link in links:
        title = link.get("title")
        href = link.get("href")

        if title and href:
            data.append({
                "judul_tempat": title,
                "url_tempat": href
            })

print("Total data ditemukan:", len(data))

for i, item in enumerate(data, start=1):
    print(i, item)
```

```python
df = pd.DataFrame(data)

display(df)
```

```python
df.to_csv(
    "data/raw/data_wisata_bali.csv",
    index=False,
    encoding="utf-8-sig"
)

print("Data berhasil disimpan!")
```
