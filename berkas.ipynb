```python
import pandas as pd
import os
```

```python
# folder dataset
file_path = "data/raw/bali_tourist.csv"
# Membaca dataset CSV
df = pd.read_csv(file_path)
# Tampilkan data
display(df)
```

```python
# nama seluruh kolom
print("Daftar Kolom:")
print(df.columns.tolist())
# ukuran dataset
print("Ukuran Dataset:", df.shape)
```

```python
# aggregat kolom
df_area_year = df.groupby(['year', 'area'], as_index=False).sum(numeric_only=True)
# tampilkan
display(df_area_year)
```

```python
output_path = "data/process/bali_tourist_aggregated.csv"
# print(output_path);

# Simpan data Process
df_area_year.to_csv(output_path, index=False)
print("Dataset berhasil disimpan:", output_path)
```
